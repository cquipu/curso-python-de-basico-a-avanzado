```python
import csv, math
```

```python
def cargar_datos(filepath):
    with open(filepath, mode="r") as file:
        csv_reader=csv.reader(file)
        dataset=[row for row in csv_reader if row] # con el condicional row eliminamos las filas vacias
    return dataset
```

```python
def agrupar_especie(datos):
    agrupados = {}
    for fila in datos:
        especie = fila[-1]
        # convierte a float todas menos la especie
        caracteristicas = list(map(float, fila[:-1]))
        if especie not in agrupados:
            agrupados[especie] = {
                'longitud_sepalo': [],
                'ancho_sepalo': [],
                'longitud_petalo': [],
                'ancho_petalo': []
            }

        agrupados[especie]['longitud_sepalo'].append(caracteristicas[0])
        agrupados[especie]['ancho_sepalo'].append(caracteristicas[1])
        agrupados[especie]['longitud_petalo'].append(caracteristicas[2])
        agrupados[especie]['ancho_petalo'].append(caracteristicas[3])

    return agrupados
```

```python
class Flor:

    def __init__(self, longitud_sepalo, ancho_sepalo, longitud_petalo, ancho_petalo):
        self.longitud_sepalo = longitud_sepalo
        self.ancho_sepalo = ancho_sepalo
        self.longitud_petalo = longitud_petalo
        self.ancho_petalo = ancho_petalo

    def promedio(self, datos):
        return sum(datos)/len(datos)

    def desviacion_estandar(self, datos):
        prom = self.promedio(datos)
        varianza = sum((x-prom)**2 for x in datos)/len(datos)
        return math.sqrt(varianza)

    def analisis(self):
        analisis = {
            'longitud_sepalo': {
                'promedio': self.promedio(self.longitud_sepalo),
                'deviacion_estandar': self.desviacion_estandar(self.longitud_sepalo)
            },
            'ancho_sepalo': {
                'promedio': self.promedio(self.ancho_sepalo),
                'deviacion_estandar': self.desviacion_estandar(self.ancho_sepalo)
            },
            'longitud_petalo': {
                'promedio': self.promedio(self.longitud_petalo),
                'deviacion_estandar': self.desviacion_estandar(self.longitud_petalo)
            },
            'ancho_petalo': {
                'promedio': self.promedio(self.ancho_petalo),
                'deviacion_estandar': self.desviacion_estandar(self.ancho_petalo)
            }
        }
        return analisis
```

```python
datos = cargar_datos('iris/iris.data.txt')
flores = agrupar_especie(datos)
iris_setosa = Flor(
    flores['Iris-setosa']['longitud_sepalo'],
    flores['Iris-setosa']['ancho_sepalo'],
    flores['Iris-setosa']['longitud_petalo'],
    flores['Iris-setosa']['ancho_petalo'],
)


# print(datos)
```

```python
print(iris_setosa.analisis())
```

```python
iris_versicolor = Flor(
    flores['Iris-versicolor']['longitud_sepalo'],
    flores['Iris-versicolor']['ancho_sepalo'],
    flores['Iris-versicolor']['longitud_petalo'],
    flores['Iris-versicolor']['ancho_petalo'],
)

print(iris_versicolor.analisis())
```
