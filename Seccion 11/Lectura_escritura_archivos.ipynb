## Lectura y escritura de archivos
'r' : Lectura (por defector). Abre un archivo para lectura
'w' : Escritura. Crea un archivo si no existe o sobreescribe el archivo existente
'a' : Añadir.  Abre un archivo
'r+': Lectura y escritura
'b' : Modo Binario

```python
archivo = open('prueba.txt', 'r')
```

```python
contenido = archivo.read()
print(contenido)
archivo.close()
```

```python
archivo=open('prueba.txt','w')
archivo.write("Hola, mundo\n")
archivo.close()
```

```python
with open('prueba.txt','r') as archivo:
    contenido=archivo.read()
    print(contenido)    
```

```python
with open('prueba.txt','w') as archivo:
    archivo.write("Adios. mundo!\n")
   
```

```python
with open('prueba.txt','a') as archivo:
    archivo.write("Hola otra vez. mundo!\n")
```

## Diferentes tipos de archivos

```python
import csv
```

```python
with open('datos.csv', mode='w') as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(['nombre', 'edad', 'ciudad'])
    escritor.writerow(['Alice', '30', 'New York'])
    escritor.writerow(['Juan', '24', 'Quito'])
```

```python
with open('datos.csv', mode='r') as archivo:
    lector = csv.reader(archivo)
    for fila in lector:
        print(fila)
```

```python
import json
```

```python
with open('datos.json', mode='w') as archivo:
    json.dump({'nombre':'Alice', 'edad':30}, archivo)
```

```python
with open('datos.json', mode='r') as archivo:
    datos=json.load(archivo)
    print(datos)
```
