```python
# [nueva_expresion for elemento in iterable if condicion] -> list
cuadrados = [x**2 for x in range(10)]
print(cuadrados)
```

```python
cuadrados = [x**2 for x in range(10) if x % 2 == 0]
print(cuadrados)
```

```python
# {nueva_expresion for elemento in iterable if condicion} -> set
cuadrados_set = {x**2 for x in range(10)}
print(cuadrados_set)
```

```python
# {clave:valor for elemento in iterable if condicion} -> dict
```

```python
cuadrados_dict = {x:x**2 for x in range(1,5)}
print(cuadrados_dict)
```

```python
lista=[]
for i in range(10):
    lista.append(i**2)
print(lista)
```

```python
matriz = [[j for j in range(4)] for i in range(3)]
print(matriz)
```

```python
matriz = [[j**2 for j in range(4)] for i in range(3)]
print(matriz)
```

```python
nombres=["Ana","Bob","Carlos"]
edades=[24,30,36]

for nombre,edad in zip(nombres, edades):
    print(f'{nombre} tiene {edad} años')
    
```

```python
lista_nueva={nom:edad for nom,edad in zip(nombres, edades)}
print(lista_nueva)
```

```python
zip(nombres, edades) # iterador
```
