```python
class Contador:
    def __init__(self, bajo, alto):
        self.actual = bajo
        self.alto = alto

    def __iter__(self):
        return self

    def __next__(self):
        if self.actual < self.alto:
            num = self.actual
            self.actual += 1
            return num
        raise StopIteration
    

#Usando el iterador
for num in Contador(3,8):
    print(num)
```

## Funciones generadoras

```python
def contador(bajo,alto):
    while bajo<alto:
        yield bajo
        bajo+=1


#usando la funcion generadora
for i in contador(3,8):
    print(i)
```

## Compresion de generadores

```python
# (valor for valor in iterable if condicion)

cuadrados_pares_gen = (x**2 for x in range(10) if x % 2 == 0)
print(cuadrados_pares_gen)
```

```python
for num in cuadrados_pares_gen:
    print(num)
```
